## 1. Qëllimi i Eksperimentit

Ky laborator shqyrton mënyrën se si sillet tensioni dalës $V_{\text{out}}$ dhe raporti i tensionit $U_{\text{out}}/U_{\text{in}}$ kur ndryshon shkalla e rezistorëve ($\text{k}\Omega$ kundrejt $\text{M}\Omega$) dhe kur qarkut i shtohet një rezistencë ngarkese $R_L$[span_4](start_span)[span_4](end_span)[span_5](start_span)[span_5](end_span).

* **Sistemi i shqyrtuar:** Qark pjesëtues tensioni i testuar me çifte rezistorësh në $\text{k}\Omega$ dhe $\text{M}\Omega$, si dhe me ngarkesa të ndryshme $R_L$ ($46\ \Omega$, $100\ \Omega$, $220\ \Omega$, $470\ \Omega$ dhe $1000\ \Omega$)[span_6](start_span)[span_6](end_span)[span_7](start_span)[span_7](end_span).
* **Fenomeni i ndjekur:** Ndryshimi i raportit $U_{\text{out}}/U_{\text{in}}$ ndërmjet matjeve me rezistorë në $\text{k}\Omega$ dhe $\text{M}\Omega$, si dhe rritja e $V_{\text{out}}$ me rritjen e ngarkesës $R_L$[span_8](start_span)[span_8](end_span)[span_9](start_span)[span_9](end_span).
* **Madhësitë e regjistruara:** $R_1$, $R_2$, $R_3 / R_L$, $V_{\text{in}}$, $V_{\text{out}}$ dhe raporti $U_{\text{out}}/U_{\text{in}}$[span_10](start_span)[span_10](end_span).

## 2. Baza Teorike

Te një pjesëtues tensioni me dy rezistorë $R_1$ dhe $R_2$, raporti teorik i tensionit përcaktohet nga:
$$\frac{U_{\text{out}}}{U_{\text{in}}} = \frac{R_2}{R_1 + R_2}$$

Kur punojmë me vlera të mëdha rezistence (si ato në $\text{M}\Omega$) ose kur lidhim një ngarkesë $R_L$ në dalje, ngarkimi i qarkut dhe rezistenca e brendshme/hyrëse e pajisjeve matëse ndikojnë në vlerën përfundimtare të tensionit të lexuar.
