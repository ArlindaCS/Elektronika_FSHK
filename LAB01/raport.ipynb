**Qëllimi:** Matja dhe krahasimi i tensionit, rrymës dhe amplifikimit në konfigurimet **Burimi**, **Seri** dhe **Paralel** me rezistorët `120 kΩ` dhe `1.77 kΩ`.

```python
import csv
import io

csv_data = """Konfigurimi,Rezistenca,Rryma (A),Vin (V),Vout (V),Amplifikimi
Burimi,"120 kΩ dhe 1.77 kΩ",2.263,11.69,N/A,
Seri,"120 kΩ dhe 1.77 kΩ",0.0,11.69,3.93,0.3361847733105218
Paralel,"120 kΩ dhe 1.77 kΩ",2.256,11.69,11.36,0.9717707442258341"""

rows = list(csv.reader(io.StringIO(csv_data)))
for r in rows:
    print(" | ".join(r))
```

## Konkluzione
- **Burimi:** Tensioni hyrës $V_{in} = 11.69\text{ V}$ dhe rryma $2.263\text{ A}$.
- **Seri:** Tensioni dalës $V_{out} = 3.93\text{ V}$ me amplifikim rreth $0.336$, ndërsa rryma matet $0.0\text{ A}$.
- **Paralel:** Tensioni dalës $V_{out} = 11.36\text{ V}$ me amplifikim rreth $0.972$ dhe rrymë $2.256\text{ A}$.
