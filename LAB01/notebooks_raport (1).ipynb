## 1. Qëllimi i Eksperimentit

Në këtë eksperiment u krye matja e dy rezistencave me vlera nominale $120\text{ k}\Omega$ dhe $1.77\text{ k}\Omega$, montimi i tyre në **breadboard** dhe lidhja me furnizuesin e energjisë duke matur tensionin e burimit $11.69\text{ V}$ dhe rrymën $2.263\text{ A}$, si dhe u analizuan konfigurimet **seri** dhe **paralel**.

* **Cili qark/sistem analizohet?** Qark rezistiv me rezistorët $120\text{ k}\Omega$ dhe $1.77\text{ k}\Omega$ i lidhur në breadboard me burim tensioni DC dhe i matur me multimetër dixhital.
* **Cili parim fizik verifikohet?** Ligji i Ohm-it dhe rregullat e shpërndarjes së tensionit dhe rrymës në lidhjet seri dhe paralel.
* **Cili parametër përcaktohet?** Tensioni hyrës $V_{in}$, tensioni dalës $V_{out}$, rryma elektrike $I$ dhe amplifikimi/raporti i tensionit.
* **Cila marrëdhënie teorike testohet?** Marrëdhënia midis rezistencës ekuivalente, rrymës dhe rënies së tensionit në konfigurimet seri, paralel dhe te burimi.

## 2. Baza Teorike

### 2.1 Modeli i Qarkut
Për realizimin e matjeve u përdorën:
* **Rezistorët:** $R_1 = 120\text{ k}\Omega$ dhe $R_2 = 1.77\text{ k}\Omega$.
* **Tensioni i furnizuesit laboratorik:** $V_{in} = 11.69\text{ V}$.
* **Pajisjet matëse:** Multimetri dixhital, furnizuesi laboratorik DC dhe breadboard-i.

### 2.2 Ekuacionet Themelore
Ligji i Ohm-it jepet nga formula:
$$I = \frac{V}{R}$$

Për rezistencën ekuivalente në konfigurimet seri dhe paralel:
$$R_{\text{seri}} = R_1 + R_2, \qquad \frac{1}{R_{\text{paralel}}} = \frac{1}{R_1} + \frac{1}{R_2}$$

## 6. Analiza e Gabimeve

Llogaritja e gabimit relativ bëhet duke krahasuar vlerat eksperimentale me pritshmëritë teorike:
$$\text{Gabimi} = \frac{|V_{\text{exp}} - V_{\text{teori}}|}{V_{\text{teori}}} \times 100\%$$

* **Paralel:** Për tensionin e burimit $V_{\text{teori}} = 11.69\text{ V}$ dhe tensionin e matur $V_{\text{exp}} = 11.36\text{ V}$, devijimi absolut i tensionit është $|11.36 - 11.69| = 0.33\text{ V}$.
* **Seri:** Për tensionin e matur $V_{\text{exp}} = 3.93\text{ V}$ krahasuar me vlerën teorike të pritur $V_{\text{teori}} \approx 11.52\text{ V}$, vërehet rënie e ndjeshme e tensionit dhe rrymë $0\text{ A}$ në shkallën e përdorur.
* **Tolerancat e komponentëve:** Rezistorët mund të kenë devijime nga vlera nominale.
* **Pasiguritë dhe kufizimet e instrumenteve:** Luhatje të vogla të leximeve, rezistenca e kontakteve në breadboard dhe përçuesit lidhës.
