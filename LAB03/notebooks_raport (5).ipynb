## 1. Qëllimi i Eksperimentit

Ky laborator synon shqyrtimin e sinjaleve periodike dhe sjelljes së qarqeve me kondensatorë (`C1–C4`) nëpërmjet matjeve të periodës $T$ dhe frekuencës $f$ me oshiloskop[span_5](start_span)[span_5](end_span)[span_6](start_span)[span_6](end_span)[span_7](start_span)[span_7](end_span).

* **Sistemi i shqyrtuar:** Qark me kondensatorë (`C1–C4`) ku vëzhgohen parametrat kohorë dhe frekuencorë të sinjalit[span_8](start_span)[span_8](end_span)[span_9](start_span)[span_9](end_span)[span_10](start_span)[span_10](end_span).
* **Madhësitë e matura:** Perioda $T$ (në sekonda / mikrosekonda) dhe frekuenca $f$ (në $\text{kHz}$ dhe $\text{MHz}$) për katër kondensatorët[span_11](start_span)[span_11](end_span)[span_12](start_span)[span_12](end_span)[span_13](start_span)[span_13](end_span).
* **Lidhja fizike që ndiqet:** Marrëdhënia ndërmjet periodës $T$ dhe frekuencës $f$, si dhe dallimet midis vlerave të regjistruara në matje dhe pritshmërisë teorike $f = 1/T$[span_14](start_span)[span_14](end_span)[span_15](start_span)[span_15](end_span)[span_16](start_span)[span_16](end_span).

## 2. Baza Teorike

Për një sinjal periodik me periodë $T$, frekuenca teorike jepet nga marrëdhënia e anasjelltë:
$$f = \frac{1}{T}$$

Në qarqe RC dhe në matjet me oshiloskop, koha e karikimit/shkarkimit dhe forma e valës lidhen me konstanten kohore $\tau = RC$ dhe me periodën e sinjalit, ndërsa çdo pasiguri në leximin e shkallës kohore ose frekuencës ndikon në përputhjen midis $f_{\text{matur}}$ dhe $1/T$.

## 5. Analiza e Te Dhenave dhe Grafiket

```python
import csv
import io
import matplotlib.pyplot as plt

csv_data = """Kondesator,T_s,f_kHz
C1,3e-6,110.7
C2,6e-6,333.3
C3,2e-6,300.0
C4,1e-6,1000.0"""

kondensatoret = []
periodat_us = []
frekuencat_khz = []

reader = csv.DictReader(io.StringIO(csv_data))
for row in reader:
    kondensatoret.append(row["Kondesator"])
    periodat_us.append(float(row["T_s"]) * 1e6)
    frekuencat_khz.append(float(row["f_kHz"]))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4.5))

ax1.bar(kondensatoret, periodat_us, color=["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728"])
ax1.set_title("Perioda T sipas Kondensatorit")
ax1.set_xlabel("Kondensatori")
ax1.set_ylabel("Perioda T (µs)")
ax1.grid(True, axis="y", linestyle="--", alpha=0.6)

ax2.bar(kondensatoret, frekuencat_khz, color=["#9467bd", "#8c564b", "#17becf", "#e377c2"])
ax2.set_title("Frekuenca f sipas Kondensatorit")
ax2.set_xlabel("Kondensatori")
ax2.set_ylabel("Frekuenca f (kHz)")
ax2.grid(True, axis="y", linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```

## 6. Analiza e Gabimeve

* **Kondensatori 1 ($T = 3 \cdot 10^{-6}\text{ s}$):** Frekuenca e regjistruar është $110.7\text{ kHz}$ ($0.1107\text{ MHz}$), e cila devijon ndjeshëm nga vlera e thjeshtë $1/T \approx 333.3\text{ kHz}$).
* **Kondensatorët 2, 3 dhe 4:** Regjistrohen përkatësisht $T = 6 \cdot 10^{-6}\text{ s}$ me $f = 333.3\text{ kHz}$, $T = 2 \cdot 10^{-6}\text{ s}$ me $f = 300\text{ kHz}$ dhe $T = 1 \cdot 10^{-6}\text{ s}$ me $f = 1\text{ MHz}$ ($1000\text{ kHz}$). 
* **Burimet e pasigurisë:** Leximi i shkallës kohore/frekuencore në oshiloskop, tolerancat e vlerave të kapacitetit/rezistencës në qark, zhurma e sinjalit dhe kohëzgjatja e regjistrimit të formës së valës.
